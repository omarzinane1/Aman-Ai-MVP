```python
from queue import Full


{
 "cells": [
  {
   "cell_type": "markdown",
   "metadata": {},
   "source": [
    "# Notebook 01 – Prétraitement audio (Mel Spectrogramme)\n",
    "**Dataset Danger Detection (Women & Children)**  \n",
    "MOHATI Ahmed – Co-leader Training"
   ]
  },
  {
   "cell_type": "code",
   "execution_count": Full,
   "metadata": {},
   "outputs": [],
   "source": [
    "import librosa\n",
    "import numpy as np\n",
    "import os\n",
    "from tqdm import tqdm\n",
    "\n",
    "# ===================== CONFIGURATION PROJET =====================\n",
    "PROJECT_ROOT = os.path.abspath(os.path.join(os.getcwd(), \"..\", \"..\"))  # aman_ai_mvp/\n",
    "DATA_AUDIO   = os.path.join(PROJECT_ROOT, \"data\", \"audio\")\n",
    "\n",
    "print(\"📁 Project root  :\", PROJECT_ROOT)\n",
    "print(\"📁 Dossier audio :\", DATA_AUDIO)\n",
    "\n",
    "# ===================== FONCTION DE PRÉTRAITEMENT =====================\n",
    "def load_and_preprocess(file_path, sr=16000, duration=3):\n",
    "    y, _ = librosa.load(file_path, sr=sr, duration=duration)\n",
    "    \n",
    "    # Padding ou troncature pour avoir exactement 3 secondes\n",
    "    target_len = sr * duration\n",
    "    if len(y) < target_len:\n",
    "        y = np.pad(y, (0, target_len - len(y)))\n",
    "    else:\n",
    "        y = y[:target_len]\n",
    "    \n",
    "    # Extraction Mel-spectrogramme\n",
    "    mel = librosa.feature.melspectrogram(y=y, sr=sr, n_mels=128, n_fft=2048, hop_length=512)\n",
    "    mel_db = librosa.power_to_db(mel, ref=np.max)\n",
    "    \n",
    "    # Taille fixe attendue par le modèle CNN\n",
    "    mel_db = mel_db[:128, :124]\n",
    "    return mel_db[..., np.newaxis]  # shape (128, 124, 1)\n",
    "\n",
    "# ===================== PRÉTRAITEMENT DE TOUS LES FICHIERS =====================\n",
    "total_processed = 0\n",
    "\n",
    "for cls in ['normal', 'cry', 'fight']:\n",
    "    path = os.path.join(DATA_AUDIO, cls)\n",
    "    if not os.path.exists(path):\n",
    "        print(f\"⚠️ Dossier manquant : {path}\")\n",
    "        continue\n",
    "    \n",
    "    wav_files = [f for f in os.listdir(path) if f.endswith('.wav')]\n",
    "    print(f\"\\n🔄 Prétraitement de la classe '{cls}' → {len(wav_files)} fichiers\")\n",
    "    \n",
    "    for f in tqdm(wav_files, desc=cls):\n",
    "        wav_path = os.path.join(path, f)\n",
    "        npy_path = os.path.join(path, f.replace('.wav', '.npy'))\n",
    "        \n",
    "        mel = load_and_preprocess(wav_path)\n",
    "        np.save(npy_path, mel)\n",
    "        total_processed += 1\n",
    "\n",
    "print(f\"\\n🎉 PRÉTRAITEMENT TERMINÉ !\")\n",
    "print(f\"   → {total_processed} fichiers .npy créés\")\n",
    "print(f\"   → Classes : normal | cry | fight\")\n",
    "print(f\"\\n🚀 Prochaine étape : Exécute 02_train_audio.ipynb\")"
   ]
  }
 ],
 "metadata": {
  "kernelspec": {
   "display_name": "Python 3",
   "language": "python",
   "name": "python3"
  },
  "language_info": {
   "name": "python"
  }
 },
 "nbformat": 4,
 "nbformat_minor": 4
}
```
